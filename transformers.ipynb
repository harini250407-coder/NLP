```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "human-centered-summarization/financial-summarization-pegasus"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

input_text = "Kedarnath is one of the holiest and most remote pilgrimage sites in Hinduism. Located in the Garhwal Himalayas of Uttarakhand, India, at an altitude of 3,583 meters (11,755 feet), it is the highest of the 12 sacred Jyotirlingas and a key stop on the Chota Char Dham Yatra.Here are the most important details about the shrine:Mythological Origin: According to the epic Mahabharata, the Pandava brothers sought Lord Shiva's forgiveness for killing their kin. Shiva transformed into a bull to hide, but the Pandavas caught him. Shiva then disappeared into the ground, leaving his hump at Kedarnath.The Temple Structure: The current temple is believed to be over 1,000 years old, restored in the 8th century by the Hindu philosopher Adi Shankaracharya. Built of massive grey stone slabs interlocked without mortar, the unique triangular Shivalinga represents the bull's hump.The 2013 Miracle: In 2013, devastating flash floods and landslides swept through the region, destroying much of the township. The temple itself, however, survived largely intact, protected by an enormous boulder (the Bheem Shila) that diverted the floodwaters away from the sanctum.How to Reach: The temple is closed during the harsh winter months (November to April/May), and the deities are relocated to Ukhimath. During the open season, reaching the shrine requires a 16-kilometer (10-mile) trek or a helicopter flight from the base camp at Gaurikund.Recent Trends: The pilgrimage is immensely popular and frequently sees massive footfalls, drawing over a million devotees during peak seasons"

inputs = tokenizer(input_text, max_length=1024, truncation=True, return_tensors="pt")
summary_ids = model.generate(inputs["input_ids"], num_beams=4, max_length=150, early_stopping=True)
summarized_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print(summarized_text)
```

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

model_name = "google-bert/bert-large-cased-whole-word-masking-finetuned-squad"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)
```

```python
ner = pipeline("ner", aggregation_strategy="simple")

text = "Elon Musk founded SpaceX and Tesla in the United States."

result = ner(text)

for entity in result:
    print(f"Entity: {entity['word']}, Type: {entity['entity_group']}")
```
